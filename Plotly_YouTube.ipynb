```python
import matplotlib.pyplot as plt

import seaborn as sns

import plotly.express as px
```

```python
import plotly.graph_objects as go
```

```python
x=[1,2,4,6,8,10]
y=[3,5,7,9,11,13]

```

```python
plt.plot(x,y,color='r')
```

```python
fig=sns.lineplot(x=x,y=y,color='r')
fig
plt.show()
```

```python
fig=px.line (x=x,y=y,title='Line Graph')

fig.show()
```

```python
fig=go.Figure(go.Scatter(x=x,y=y))

fig.show()
```

# Adding Titles

```python

fig= go.Figure(go.Scatter(x=x,y=y))
fig.update_layout(title='Line-Graph',xaxis_title='This is x-axis',yaxis_title='This is y-axis')
fig.show()
```

```python
x1,y1=[1,2,3,4,5],[4,5,6,7,2]
x2,y2=[1,2,3,4,5],[4,3,6,2,1]
x3,y3=[1,2,3,4,5],[8,6,8,2,6]
```

```python
fig=go.Figure()
```

```python
fig.add_trace(go.Scatter(x=x1,y=y1))
fig.add_trace(go.Scatter(x=x2,y=y2))
fig.add_trace(go.Scatter(x=x3,y=y3))

fig.show()
```

```python
fig=go.Figure()
```

```python
fig.add_trace(go.Scatter(x=x1,y=y1, name='line1',mode='lines'))
fig.add_trace(go.Scatter(x=x2,y=y2, name='line2',mode='markers'))
fig.add_trace(go.Scatter(x=x3,y=y3, name='line3',mode='lines+markers'))

fig.show()
```

```python
fig = px.scatter(x=x, y=y)
fig.show()
```

# Bubble Charts

```python
x=[1,2,3,4]
y=[4,6,8,10]

fig=go.Figure(go.Scatter(x=x,y=y,mode='markers'))

fig.show()
```

```python
x=[1,2,3,4]
y=[4,6,8,10]

fig=go.Figure(go.Scatter(x=x,y=y,mode='markers',marker_size=[40,60,70,80]))

fig.show()
```

```python
x=[1,2,3,4]
y=[4,6,8,10]

fig=go.Figure(go.Scatter(x=x,y=y,mode='markers',marker_size=[40,60,70,80],
            text=['product A','product B','product C','product D']))

fig.show()
```

```python
d =px.data.tips().query("total_bill > 15.00")
```
